```python
import json
import matplotlib.pyplot as plt
```

```python
runs = {
    "frozen": "../reports/frozen.json",
    "FT lr 1e-4": "../reports/fine_tuned.json",
    "FT lr 3e-5": "../reports/history_ft_lr_3e-5.json",
    "FT lr 1e-5": "../reports/history_ft_lr_1e-5.json"
}
fig, (ax_loss, ax_acc) = plt.subplots(1, 2, figsize=(12, 4))

for name, path in runs.items():
    with open(path) as f:
        h = json.load(f)
    epochs = range(1, len(h["val_loss"]) + 1)
    ax_loss.plot(epochs, h["val_loss"], marker="o", label=name)
    ax_acc.plot(epochs, h["val_acc"], marker="o", label=name)

ax_loss.set(title="Validation loss", xlabel="Epoch", ylabel="Cross-entropy loss")
ax_acc.set(title="Validation accuracy", xlabel="Epoch", ylabel="Accuracy")

for ax in (ax_loss, ax_acc):
    ax.legend()
    ax.grid(alpha=0.3)
    ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))

plt.tight_layout()
plt.savefig("../reports/learning_curves.png", dpi=150)
plt.show()
```

##### Validation loss and acccuracy per epoch
the frozen run trains only the classification head from a random initialisation(15 epochs);
the fine-tuning runs start from trained frozen model and update all layers (10 epochs), so their opoch axes are not directly comparable
